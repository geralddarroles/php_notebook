### PHP Data type
PHP data types are commonly grouped as:

- **Scalar types**
	- `bool` (Boolean)
	- `int` (Integer)
	- `float` (Double)
	- `string`

- **Compound types**
	- `array`
	- `object`

- **Special types**
	- `null`
	- `resource`

Additional type declarations used in modern PHP include: `callable`, `iterable`, `mixed`, `void`, `never`, `false`, `true`, and union/intersection types.

## Definitive Guide to PHP

PHP is a server-side scripting language primarily used for web development. It powers dynamic websites, APIs, server-side logic, and content management systems. PHP is embedded directly in HTML and runs on the server, generating output sent to the browser.

### 3. Variables and Data Types
PHP variables begin with `$`.

```php
<?php
$name = "Alice";
$age = 30;
$price = 49.99;
$active = true;
$nothing = null;
?>
```

#### Common data types
- `int` — integers
- `float` — floating-point numbers
- `string` — text
- `bool` — boolean (`true` / `false`)
- `array` — ordered or associative collections
- `object` — instances of classes
- `null` — no value
- `resource` — special external resource
- `callable` — function/callable reference
- `iterable` — arrays or traversable objects

#### Type declarations
```php
<?php
function greet(string $name): string {
	return "Hello, $name!";
}

echo greet("Alice");
?>
```

### 4. Strings
```php
<?php
$message = "Hello";
$name = "PHP";

echo $message . " " . $name; // concatenation
echo "Welcome, $name!";      // interpolation
echo 'Welcome, $name!';      // no interpolation in single quotes
?>
```

#### Common string functions
```php
<?php
$str = "  Hello PHP  ";

trim($str);          // removes spaces
strlen($str);        // length
strtoupper($str);    // uppercase
strtolower($str);    // lowercase
strpos("hello", "el"); // find substring position
substr("Hello", 1, 4); // "ello"
?>
```

### 5. Arrays
PHP arrays can be indexed or associative.

```php
<?php
$numbers = [1, 2, 3, 4];
$colors = ["red" => "#f00", "blue" => "#00f"];
$names = ["Alice", "Bob", "Charlie"];

echo $numbers[0];
echo $colors["red"];
?>
```

#### Array functions
```php
<?php
$numbers = [3, 1, 2];

sort($numbers);         // sorts ascending
rsort($numbers);        // sorts descending
count($numbers);        // number of elements
array_push($numbers, 4); // append element
array_pop($numbers);     // remove last element
implode(", ", $numbers); // join array into string
explode(", ", "A,B,C");  // split string into array
?>
```

### 6. Operators
PHP supports:
- Arithmetic: `+`, `-`, `*`, `/`, `%`
- Comparison: `==`, `===`, `!=`, `!==`, `<`, `>`, `<=`, `>=`
- Logical: `&&`, `||`, `!`
- Assignment: `=`, `+=`, `-=`, `*=`, `/=`
- Null coalescing: `??`
- Spaceship: `<=>` (comparison returning -1, 0, 1)

```php
<?php
$a = 10;
$b = 5;

echo $a + $b;      // 15
echo $a == $b;     // false
echo $a ?? $b;     // 10 if $a is not null
?>
```

### 7. Control Structures
#### If / Else
```php
<?php
$age = 18;

if ($age >= 18) {
	echo "Adult";
} elseif ($age >= 13) {
	echo "Teen";
} else {
	echo "Child";
}
?>
```

#### Switch
```php
<?php
$day = "Tue";

switch ($day) {
	case "Mon":
		echo "Monday";
		break;
	case "Tue":
		echo "Tuesday";
		break;
	default:
		echo "Another day";
}
?>
```

#### Ternary Operator
```php
<?php
$age = 20;
$status = ($age >= 18) ? "adult" : "minor";
echo $status;
?>
```

### 8. Loops
#### For loop
```php
<?php
for ($i = 0; $i < 5; $i++) {
	echo $i . "<br>";
}
?>
```

#### While loop
```php
<?php
$i = 0;
while ($i < 5) {
	echo $i . "<br>";
	$i++;
}
?>
```

#### Foreach loop
```php
<?php
$items = ["apple", "banana", "orange"];

foreach ($items as $item) {
	echo $item . "<br>";
}
?>
```

### 9. Functions
Functions allow code reuse.

```php
<?php
function add($a, $b) {
	return $a + $b;
}

echo add(3, 4); // 7
?>
```

#### Default parameters
```php
<?php
function greet($name = "Guest") {
	return "Hello, $name!";
}

echo greet();
echo greet("Alice");
?>
```

#### Variadic functions
```php
<?php
function sum(...$numbers) {
	return array_sum($numbers);
}

echo sum(1, 2, 3, 4); // 10
?>
```

### 10. Include and Require
PHP supports including external files.

```php
<?php
include "header.php";
require "config.php";
?>
```

- `include` — produces a warning if file is missing
- `require` — throws a fatal error if file is missing

### 14. Object-Oriented Programming (OOP)
PHP supports classes and objects.

```php
<?php
class Person {
	public $name;
	public $age;

	public function __construct($name, $age) {
		$this->name = $name;
		$this->age = $age;
	}

	public function intro() {
		return "Hi, I'm {$this->name} and I'm {$this->age}.";
	}
}

$person = new Person("Alice", 30);
echo $person->intro();
?>
```

#### Inheritance
```php
<?php
class Employee extends Person {
	public $role;

	public function __construct($name, $age, $role) {
		parent::__construct($name, $age);
		$this->role = $role;
	}
}
?>
```

#### Encapsulation
```php
<?php
class BankAccount {
	private $balance = 0;

	public function deposit($amount) {
		$this->balance += $amount;
	}

	public function getBalance() {
		return $this->balance;
	}
}
?>
```

### 15. Namespaces
Namespaces prevent class name collisions.

Usage:
```php
<?php
use App\Models\User;

$user = new User();
?>
```

### 16. Error Handling
#### Try / Catch
```php
<?php
try {
	throw new Exception("Something went wrong");
} catch (Exception $e) {
	echo "Error: " . $e->getMessage();
}
?>
```

#### Custom error handler
```php
<?php
set_error_handler(function($severity, $message, $file, $line) {
	throw new ErrorException($message, 0, $severity, $file, $line);
});
?>
```

### 17. File Handling
```php
<?php
$file = fopen("data.txt", "w");
fwrite($file, "Hello PHP");
fclose($file);

$content = file_get_contents("data.txt");
echo $content;
?>
```

### 18. Database Access
PHP commonly works with MySQL using PDO.

```php
<?php
$pdo = new PDO("mysql:host=localhost;dbname=testdb", "root", "");

$stmt = $pdo->query("SELECT * FROM users");
while ($row = $stmt->fetch(PDO::FETCH_ASSOC)) {
	print_r($row);
}
?>
```

#### Prepared statements
```php
<?php
$stmt = $pdo->prepare("SELECT * FROM users WHERE id = :id");
$stmt->execute(["id" => 1]);
$user = $stmt->fetch();
?>
```

### 20. Modern PHP Features
Modern PHP versions include useful features:

- Arrow functions

#### Match expression
```php
<?php
$day = 2;
$result = match ($day) {
	1 => "Monday",
	2 => "Tuesday",
	default => "Another day",
};

echo $result;
?>
```

#### Arrow function
```php
<?php
$numbers = [1, 2, 3];
$double = fn($n) => $n * 2;

print_r(array_map($double, $numbers));
?>
```
